```python
cols_to_fix = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
               'TechSupport', 'StreamingTV', 'StreamingMovies']

for col in cols_to_fix:
    df[col] = df[col].replace('No internet service', 'No')

df['MultipleLines'] = df['MultipleLines'].replace('No phone service', 'No')
```

```python
binary_cols = ['gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines',
               'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport',
               'StreamingTV', 'StreamingMovies', 'PaperlessBilling', 'Churn']

for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0, 'Male': 1, 'Female': 0})
```

## Data Cleaning Summary

Cleaned the raw Telco churn dataset in preparation for modeling:

- **Fixed `TotalCharges`**: 11 rows had blank values due to `tenure = 0` (brand new customers 
  with no billing history yet). Replaced with `0` and converted the column from text to numeric, 
  rather than dropping these rows. Model should handle bran new customers correctly. 
- **Dropped `customerID`**: a unique, non predictive identifier that risked encouraging the model 
  to memorize individual rows rather than learn generalizable patterns.
- **Simplified redundant categories**: collapsed "No internet service" / "No phone service" labels 
  down to plain "No" across 7 columns, since that information was already fully captured by 
  `InternetService`/`PhoneService`.
- **Encoded categorical variables**: binary Yes/No (and gender) columns mapped to 0/1; multi category 
  columns (`InternetService`, `Contract`, `PaymentMethod`) one hot encoded with `drop_first=True` 
  to avoid redundant, perfectly correlated columns.

Final dataset: 7043 rows × 24 columns, fully numeric and ready for train/test splitting.
